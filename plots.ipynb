```python
# Import plotting functions
import numpy as np
import matplotlib.pyplot as plt
import getdist.plots as gdplt
from getdist import loadMCSamples

# path info
import os
print(os.getcwd())
```

```python
# chain prefix
chain0 = 'chains/CMB-SPA/CMB-SPA'      # CMB-SPA
chain1 = 'chains/PPS+DESI/PPS+DESI'    # PPS + DESI 
chain2 = 'chains/PP+DESI/PP+DESI'      # PP + DESI
chain3 = 'chains/CMB-SPA+PPS+DESI'     # CMB-SPA + PPS + DESI
chain4 = 'chains/CMB-SPA+PP+DESI'      # CMB-SPA + PP + DESI

chains = np.array([chain0, chain1, chain2, chain3, chain4])
labels = ['CMB-SPA', 'PPS + DESI', 'PP + DESI', 'CMB-SPA + PPS + DESI', 'CMB-SPA + PP + DESI']
```

```python
# progress check
progress = np.genfromtxt(chain0+".progress", names=True)

steps = progress['N']
```

```python
# parameters
burn_in = 0.3

# load samples
#samples0 = loadMCSamples(chain0, settings={'ignore_rows':burn_in})
#samples1 = loadMCSamples(chain1, settings={'ignore_rows':burn_in})
#samples2 = loadMCSamples(chain2, settings={'ignore_rows':burn_in})
#samples3 = loadMCSamples(chain3, settings={'ignore_rows':burn_in})
#samples4 = loadMCSamples(chain4, settings={'ignore_rows':burn_in})

chain_selector = [0,1,2]

samples = [loadMCSamples(chains[n], settings={'ignore_rows':burn_in}) for n in chain_selector]

# show stats
#stats0 = samples0.getMargeStats()
#stats1 = samples1.getMargeStats()
#stats2 = samples2.getMargeStats()
#stats3 = samples3.getMargeStats()
#stats4 = samples4.getMargeStats()

# print stats
#print(stats1.parWithName('H0').mean)

# plot posteriors
g = gdplt.get_subplot_plotter()
g.triangle_plot(samples, 
                ['H0','ombh2', 'omch2', 'w', 'wa'], 
                filled=False, 
                title_limit=0,
                legend_labels=[labels[n] for n in chain_selector])

plt.legend()
plt.show()
```
